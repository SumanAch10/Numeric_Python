```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Create the dataset
X = np.array([1000,2500,3500,4000])
y = np.array([150,270,280,300])

w = 0.9
b = 0.3

fig, ax = plt.subplots(figsize = (8,5))
ax.set_ylabel("Housing_cost in dollars")
ax.set_xlabel("Area in sqft")

def predict(X,w,b):
    y_predict = np.dot(X,w) + b
    return y_predict

def compute_loss(X,w,b,y):
    m = X.shape[0]
    total_loss = 0
    for i in range(m):
        total_loss = total_loss + (np.dot(X[i],w) + b  - y[i])**2
    total_loss = total_loss / (2*m)
    return total_loss

# y_predict = predict(X,w,b)
# print(y_predict)
total_loss = compute_loss(X,w,b,y)
print(total_loss)

```
